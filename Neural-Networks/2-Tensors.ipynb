## Tensors in Pytorch

A Pytorch Tensor is basically the same as a NumPy array. This means it does not know anything about deep learning or computational graphs or gradients and is just a generic n-dimensional array to be used for arbitrary numeric computation. However, the biggest difference between a NumPy array and a PyTorch Tensor is that a PyTorch Tensor can run on either CPU or GPU. To run operations on the GPU, just cast the Tensor to a cuda datatype using:

```python
device = torch.device("cpu")
```


To create random input and output data, where `N` is batch size, `D_in` is input dimension, `H` is hidden dimension, and `D_out` is output dimension:

```python
N, D_in, H, D_out = 32, 100, 10, 2

x = torch.randn(N, D_in, device=device, dtype=torch.float)    #where x is a tensor
```

In the above example, `x` can be thought of as a random feature tensor as an input to a model. We will see how to create tensors, different attributes, and operations on a tensor in this article.

## <br>How to create a Tensor?
You can create a tensor using some simple lines of code as shown below.

````python
import torch

V_data = [1, 2, 3, 4, 5]
V = torch.tensor(V_data)

print(V)

# You can also create a tensor of random data with a given dimensionality like:
x = torch.randn(3, 4, 5)
x
````

````python
# torch.zeros(): Creates a new Tensor with all elements, initialized as zeros
z = torch.zeros([3, 3], dtype=torch.int32)
print(z)

# toch ones
z = torch.ones([3, 3])
print(z)
````

- torch.full() and torch.full_like():  These functions return a Tensor of the required size filled with required fill_value provided. The complete prototype for torch.full() is:

````python
# example of torch.full()
newTensor= torch.full((4, 3), 3.14, dtype= torch.float32)
print(newTensor)

# Example for torch.full_like()
x = torch.full_like(newTensor, 3.24, dtype=None)
print(x)
````

## Tensor Attributes

Every `torch.Tensor` has three core attributes: **dtype**, **device**, and **layout**.

---

### `torch.dtype` — Data Type
Defines what kind of numbers the tensor holds (e.g. `float32`, `int64`, `bool`). PyTorch supports 12 data types in total.

---

### `torch.device` — Where it lives
Specifies whether the tensor is on **CPU** or **GPU (CUDA)**.

```python
torch.device('cuda:0')  # GPU #0
# → device(type='cuda', index=0)

torch.device('cpu')     # CPU
```

> If no device index is given, it refers to the **currently active device**.

---

### `torch.layout` — Memory Layout
Controls how tensor data is physically stored in memory. Two options:

#### 1. `torch.strided` *(default — dense tensors)*
The standard layout. **Strides** define how many memory steps to take to move to the next element in each dimension.

```python
x = torch.Tensor([[1, 2, 3, 4],
                  [5, 7, 8, 9]])
x.stride()  # → (4, 1)
```
- Next **row** → jump **4** elements
- Next **column** → jump **1** element

#### 2. `torch.sparse_coo_tensor` *(sparse tensors)*
Efficient for tensors with mostly zeros. Stores only **non-zero values** and their **indices** (COO = Coordinate format).

```python
i = [[0, 1, 1],   # row indices
     [2, 0, 2]]   # col indices
v = [3, 4, 5]     # non-zero values

s = torch.sparse_coo_tensor(i, v, (2, 3))
# tensor(indices=tensor([[0, 1, 1],
#                        [2, 0, 2]]),
#        values=tensor([3, 4, 5]),
#        size=(2, 3), nnz=3, layout=torch.sparse_coo)
```
A 2×3 matrix with only 3 stored values instead of 6 — saves memory when most elements are zero.

## <br>Tensor operations:

````python
# 1. You can add two tensors like matrix addition.
x = torch.tensor([1., 2., 3.])
y = torch.tensor([4., 5., 6.])
z = x + y

print(z)


# 2. Concatenates a list of tensors
x_1 = torch.randn(2, 5)
y_1 = torch.randn(3, 5)
z_1 = torch.cat([x_1, y_1])

print(z_1)


# 3. To concatenate columns you can do the following
x_2 = torch.randn(2, 3)
y_2 = torch.randn(2, 5)

# second argument specifies which axis to concat along
z_2 = torch.cat([x_2, y_2], 1)
print(z_2)


# 4. You can reshape tensors using the .view() method as shown below.
x = torch.randn(2, 3, 4)
print(x)

# reshape to 2 rows, 12 columns
print(x.view(2, 12)) 


# 5. Returns the index of the maximum value of all elements in the input tensor.
x = torch.randn(3,3)
print((x, torch.argmax(x)))


# 6. Similar to argmax(), it returns the minimum value of all elements in the input tensor.
x = torch.randn(3,3)
print((x, torch.argmin(x)))
````
