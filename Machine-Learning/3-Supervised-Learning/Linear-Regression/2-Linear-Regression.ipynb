## <br>Example: Predicting Exam Scores

Consider predicting a student's exam score based on study hours. As study hours increase, scores tend to go up.

- **Independent variable (input)**: Hours studied - the factor we control or observe
- **Dependent variable (output)**: Exam score - depends on hours studied

We use the independent variable to predict the dependent variable.

## <br>Best Fit Line in Linear Regression

The best-fit line is the straight line that most accurately represents the relationship between independent and dependent variables, minimizing the difference between actual and predicted values.

![Observed Values](img/observed_value.webp)

### 1. Goal of the Best-Fit Line

Find a line that minimizes the error between observed data points and predicted values, enabling predictions for new, unseen data.

**Y** is the dependent or target variable  
**X** is the independent variable (predictor)

- $\theta_1$ represents the intercept (value of Y when X = 0)
- $\theta_2$ represents the slope (how much Y changes for a unit change in X)

### 2. Equation of the Best-Fit Line

For simple linear regression:

$$y = mx + b$$

Where:
- $y$ is the predicted value (dependent variable)
- $x$ is the input (independent variable)
- $m$ is the slope (how much y changes when x changes)
- $b$ is the intercept (value of y when x = 0)

The best-fit line optimizes m and b to minimize prediction error.

### 3. Minimizing Error: The Least Squares Method

The **Least Squares** method minimizes the sum of squared differences between actual and predicted values (residuals).

**Residual:**

$$\text{Residual} = y_i - \hat{y}_i$$

Where:
- $y_i$ is the actual observed value
- $\hat{y}_i$ is the predicted value from the line

**Minimize:**

$$\sum(y_i - \hat{y}_i)^2$$

This ensures the line best represents the data with minimal squared differences.

### 4. Interpretation of the Best-Fit Line

**Slope (m):** Indicates how much y changes per unit change in x. For example, slope = 5 means y increases by 5 units for every 1-unit increase in x.

**Intercept (b):** The predicted y value when x = 0; where the line crosses the y-axis.

**Limitations:**
- **Assumes Linearity:** Won't work well for non-linear relationships
- **Sensitivity to Outliers:** Outliers can significantly skew the line

## <br>Hypothesis Function in Linear Regression

The hypothesis function represents the relationship between input features and target output.

**Simple Linear Regression (one variable):**

$$h(x) = \beta_0 + \beta_1 x$$

Where:
- $h(x)$ or $\hat{y}$ is the predicted value
- $x$ is the independent variable
- $\beta_0$ is the intercept
- $\beta_1$ is the slope

**Multiple Linear Regression:**

$$h(x_1, x_2, ..., x_k) = \beta_0 + \beta_1 x_1 + \beta_2 x_2 + ... + \beta_k x_k$$

Where:
- $x_1, x_2, ..., x_k$ are independent variables
- $\beta_0$ is the intercept
- $\beta_1, \beta_2, ..., \beta_k$ are coefficients representing each variable's influence

## <br>Evaluation Metrics for Linear Regression

A variety of evaluation measures can be used to assess the strength of a linear regression model. These metrics indicate how well the model fits the observed data.

### 1. Mean Squared Error (MSE)
Mean Squared Error computes the average of the squared differences between actual and predicted values. Squaring ensures negative and positive errors do not cancel out.

$$
\mathrm{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

**Where:**
- $n$ is the number of data points
- $y_i$ is the actual value for the $i$-th data point
- $\hat{y}_i$ is the predicted value for the $i$-th data point

MSE is sensitive to outliers because large errors are heavily penalized.

### 2. Mean Absolute Error (MAE)
Mean Absolute Error measures the average absolute difference between predicted and actual values.

$$
\mathrm{MAE} = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|
$$

**Where:**
- $n$ is the number of observations
- $y_i$ is the actual value
- $\hat{y}_i$ is the predicted value

Lower MAE indicates better model performance. MAE is less sensitive to outliers than MSE.

### 3. Root Mean Squared Error (RMSE)
Root Mean Squared Error is the square root of MSE. It is in the same unit as the target variable and emphasizes larger errors.

$$
\mathrm{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}
$$

**Where:**
- $n$ is the number of observations
- $y_i$ is the actual value
- $\hat{y}_i$ is the predicted value

### 4. Coefficient of Determination ($R^2$)
$R^2$ indicates how much of the variance in the dependent variable is explained by the model. It ranges from 0 to 1.

$$
R^2 = 1 - \frac{\mathrm{RSS}}{\mathrm{TSS}}
$$

**Residual Sum of Squares (RSS):**

$$
\mathrm{RSS} = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$

**Total Sum of Squares (TSS):**

$$
\mathrm{TSS} = \sum_{i=1}^{n} (y_i - \bar{y})^2
$$

$R^2$ measures the proportion of variance in $y$ explained by the independent variables.

### 5. Adjusted $R^2$
Adjusted $R^2$ accounts for the number of predictors in the model and penalizes adding irrelevant features.

$$
\mathrm{Adjusted}\ R^2 = 1 - \left(1 - R^2\right) \frac{n - 1}{n - k - 1}
$$

**Where:**
- $n$ is the number of observations
- $k$ is the number of predictors
- $R^2$ is the coefficient of determination

Adjusted $R^2$ helps prevent overfitting by discouraging unnecessary predictors.


## <br>Regularization Techniques for Linear Models

Regularization is a technique used to prevent overfitting in machine learning models by adding a penalty term to the objective function. This discourages the model from learning overly complex patterns and helps improve generalization to unseen data.

### Overview of Terms Used in Regularization

Before diving into specific techniques, let's clarify the notation:

- **m**: The number of training samples (data points) in your dataset
- **n**: The number of features (independent variables) in the model
- **$\theta_j$**: The regression coefficients for each feature
- **$\hat{y}_i$**: The predicted value for the i-th sample
- **$y_i$**: The actual observed value for the i-th sample
- **$\lambda$ (lambda)**: The regularization strength hyperparameter that controls how much the model penalizes large coefficients. Higher $\lambda$ means stronger regularization
- **$\frac{1}{2m}$**: A scaling factor that normalizes the loss across all training samples. The division by m ensures the loss doesn't grow larger as your dataset gets bigger, making the regularization parameter $\lambda$ comparable across different dataset sizes

---

### 1. Lasso Regression (L1 Regularization)

Lasso Regression is a regularization technique that adds a penalty term based on the absolute values of the regression coefficients. It penalizes the sum of absolute values of coefficients, which can shrink some coefficients to exactly zero, effectively performing feature selection.

**Objective Function:**

$$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}_i - y_i)^2 + \lambda \sum_{j=1}^{n} |\theta_j|$$

**Components:**
- The **first term** is the least squares loss, representing the average squared difference between predicted and actual values (scaled by $\frac{1}{2m}$)
- The **second term** is the L1 regularization penalty, which penalizes the sum of absolute values of the regression coefficients $\theta_j$

**Key Advantages:**
- Can perform automatic feature selection by shrinking irrelevant coefficients to exactly zero
- Useful when you suspect many features are not important
- Results in sparse models (fewer non-zero coefficients)

---

### 2. Ridge Regression (L2 Regularization)

Ridge Regression is a linear regression technique that adds a penalty based on the squared values of the regression coefficients. It penalizes large coefficients more heavily than Lasso, but unlike Lasso, it rarely shrinks coefficients exactly to zero. Ridge regression is particularly useful when your dataset exhibits multicollinearity (when predictor variables are highly correlated with each other).

**Objective Function:**

$$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}_i - y_i)^2 + \lambda \sum_{j=1}^{n} \theta_j^2$$

**Components:**
- The **first term** is the least squares loss, representing the average squared difference between predicted and actual values
- The **second term** is the L2 regularization penalty, which penalizes the sum of squared values of the regression coefficients $\theta_j$

**Key Advantages:**
- Handles multicollinearity well by distributing weights among correlated features
- More stable than ordinary least squares when features are correlated
- Typically produces better generalization than unregularized linear regression

---

### 3. Elastic Net Regression

Elastic Net Regression is a hybrid regularization technique that combines both L1 (Lasso) and L2 (Ridge) regularization. It provides a balance between feature selection (from L1) and handling multicollinearity (from L2), making it versatile for many real-world problems.

**Objective Function:**

$$J(\theta) = \frac{1}{2m} \sum_{i=1}^{m} (\hat{y}_i - y_i)^2 + \alpha\lambda \sum_{j=1}^{n} |\theta_j| + \frac{1}{2}(1-\alpha)\lambda \sum_{j=1}^{n} \theta_j^2$$

**Components:**
- The **first term** is the least squares loss
- The **second term** is the L1 regularization penalty (Lasso component)
- The **third term** is the L2 regularization penalty (Ridge component)

**Understanding the Parameters:**

- **$\lambda$ (lambda)**: The overall regularization strength. Controls how much the model penalizes large coefficients. Higher values mean stronger regularization
- **$\alpha$ (alpha)**: The mixing parameter that controls the balance between L1 and L2 regularization
  - When $\alpha = 1$: Elastic Net behaves like Lasso (pure L1 regularization)
  - When $\alpha = 0$: Elastic Net behaves like Ridge (pure L2 regularization)
  - When $0 < \alpha < 1$: Elastic Net combines both techniques

**Why the formula $\frac{1}{2}(1-\alpha)\lambda \sum_{j=1}^{n} \theta_j^2$?**

The $\frac{1}{2}$ factor is included to ensure consistency with Ridge regression's formulation. When $\alpha = 0$, the Elastic Net equation reduces to the standard Ridge regression equation. The $(1-\alpha)$ term ensures that:
- As $\alpha$ increases (moving toward L1), the L2 penalty decreases proportionally
- As $\alpha$ decreases (moving toward L2), the L1 penalty decreases while the L2 penalty increases
- The two penalties are inversely proportional, maintaining a balanced trade-off

**Key Advantages:**
- Combines benefits of both Lasso and Ridge regression
- Performs feature selection like Lasso while handling multicollinearity like Ridge
- Works well with high-dimensional datasets with many correlated features
- More stable than Lasso when features are highly correlated

## <br>Why Linear Regression is Important

Linear regression is fundamental to machine learning and statistics for several reasons:

**Key Importance:**
- **Simplicity & Interpretability**: Easy to understand and a great starting point for learning machine learning
- **Predictive Power**: Enables forecasting in finance, healthcare, marketing, and many other fields
- **Foundation for Advanced Models**: Concepts underpin logistic regression, neural networks, and other algorithms
- **Computational Efficiency**: Fast to train, even on large datasets, making it suitable for real-time applications
- **Insights**: Reveals relationships between variables and how one variable influences another

---

### Advantages

- **Interpretability**: Coefficients directly show how much the dependent variable changes for a one-unit change in each independent variable
- **Efficiency**: Computationally fast and handles large datasets effectively
- **Robustness**: Relatively robust to outliers compared to other algorithms
- **Baseline Model**: Serves as a good benchmark for comparing more complex algorithms
- **Widely Available**: Well-established with extensive support in libraries and software packages

### Limitations

- **Linearity Assumption**: Assumes a linear relationship; fails if data has non-linear patterns
- **Multicollinearity Sensitivity**: Highly correlated features can inflate coefficient variance and destabilize predictions
- **Feature Engineering Required**: May need to transform features into suitable formats before use
- **Overfitting & Underfitting**: Can both overfit (memorizing training data) and underfit (too simple to capture patterns)
- **Limited Complexity**: Struggles with complex relationships that require more advanced techniques

## <br>Implementation: R², Adjusted R², Ridge, Lasso, and Elastic Net

### Sample Data Overview

We use a simple dataset with 4 samples and 2 features to demonstrate all regression techniques. Here's the breakdown:

**Sample Data:**

| Sample | Feature 1 | Feature 2 | Target (y) |
|--------|-----------|-----------|-----------|
| 1      | 1         | 2         | 2         |
| 2      | 2         | 3         | 4         |
| 3      | 3         | 4         | 5         |
| 4      | 4         | 5         | 4         |

**Explanation:**
- **X (Features)**: A 4×2 matrix containing independent variables. Each row is a sample, each column is a feature.
- **y (Target)**: A vector of 4 values representing the dependent variable we want to predict.
- The model learns to identify patterns in X that predict y.

**Regression Models Explained:**
1. **Linear Regression**: Standard OLS without regularization
2. **R² (R-Squared)**: Measures how much variance is explained by the model (0 to 1 scale)
3. **Adjusted R²**: R² adjusted for the number of features to penalize overfitting
4. **Ridge Regression**: L2 regularization - shrinks coefficients but rarely to zero
5. **Lasso Regression**: L1 regularization - can shrink coefficients exactly to zero (feature selection)
6. **Elastic Net**: Combines L1 and L2 - balances feature selection with coefficient shrinkage

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.metrics import r2_score

# Sample data
X = np.array([[1, 2], [2, 3], [3, 4], [4, 5]])
y = np.array([2, 4, 5, 4])

# Display sample data as a table
df = pd.DataFrame(X, columns=['Feature 1', 'Feature 2'])
df['Target (y)'] = y
df.index.name = 'Sample'
print("Sample Data:")
print(df)
print(f"\nDataset shape - X: {X.shape}, y: {y.shape}")
```

```python
# ============ R Squared ============
model = LinearRegression()
model.fit(X, y)
r2 = r2_score(y, model.predict(X))
print(f"\n{'='*50}")
print(f"R² = {r2:.4f}")
print(f"Explanation: {r2*100:.2f}% of variance explained by the model")

# ============ Adjusted R Squared ============
n = X.shape[0]  # number of samples
p = X.shape[1]  # number of features
adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
print(f"\n{'='*50}")
print(f"Adjusted R² = {adj_r2:.4f}")
print(f"Explanation: Adjusted metric accounting for {p} features and {n} samples")

# ============ Ridge Regression ============
ridge = Ridge(alpha=1.0)
ridge.fit(X, y)
ridge_pred = ridge.predict(X)
ridge_r2 = r2_score(y, ridge_pred)
print(f"\n{'='*50}")
print(f"Ridge Regression (alpha=1.0)")
print(f"Predictions: {ridge_pred}")
print(f"R² Score: {ridge_r2:.4f}")
print(f"Coefficients: {ridge.coef_}")

# ============ Lasso Regression ============
lasso = Lasso(alpha=0.1)
lasso.fit(X, y)
lasso_pred = lasso.predict(X)
lasso_r2 = r2_score(y, lasso_pred)
print(f"\n{'='*50}")
print(f"Lasso Regression (alpha=0.1)")
print(f"Predictions: {lasso_pred}")
print(f"R² Score: {lasso_r2:.4f}")
print(f"Coefficients: {lasso.coef_}")

# ============ Elastic Net Regression ============
elasticnet = ElasticNet(alpha=0.1, l1_ratio=0.5)  # l1_ratio: 0=Ridge, 1=Lasso
elasticnet.fit(X, y)
en_pred = elasticnet.predict(X)
en_r2 = r2_score(y, en_pred)
print(f"\n{'='*50}")
print(f"Elastic Net Regression (alpha=0.1, l1_ratio=0.5)")
print(f"Predictions: {en_pred}")
print(f"R² Score: {en_r2:.4f}")
print(f"Coefficients: {elasticnet.coef_}")

# ============ Summary Comparison ============
print(f"\n{'='*50}")
print("Model Comparison Summary:")
print(f"{'='*50}")
comparison_df = pd.DataFrame({
    'Model': ['Linear Regression', 'Ridge', 'Lasso', 'Elastic Net'],
    'R² Score': [r2, ridge_r2, lasso_r2, en_r2],
    'Coef 1': [model.coef_[0], ridge.coef_[0], lasso.coef_[0], elasticnet.coef_[0]],
    'Coef 2': [model.coef_[1], ridge.coef_[1], lasso.coef_[1], elasticnet.coef_[1]]
})
print(comparison_df.to_string(index=False))
```
