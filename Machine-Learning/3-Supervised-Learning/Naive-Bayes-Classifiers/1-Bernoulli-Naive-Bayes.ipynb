# Bernoulli Naive Bayes

Bernoulli Naive Bayes is a classification algorithm that assumes **binary features** (presence/absence). It's commonly used for **spam detection**, **text classification**, and **sentiment analysis**.

We'll demonstrate it on an email spam dataset with 5,171 samples labeled as `ham` (0) or `spam` (1).
