## Task 7: Evaluate Model Performance

### Evaluation Metrics Explained:

- **MSE (Mean Squared Error)**: Average of squared errors. Penalizes large errors more heavily.
- **RMSE (Root Mean Squared Error)**: Square root of MSE. In same units as target variable (price).
- **MAE (Mean Absolute Error)**: Average absolute error. More intuitive than MSE.
- **R² (R-Squared)**: Coefficient of determination. Proportion of variance explained (0 to 1). Higher is better.
